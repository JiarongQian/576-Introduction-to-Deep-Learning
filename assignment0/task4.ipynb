```python
import matplotlib.pyplot as plt

labels = ["Y1", "Y2", "Y3", "Y4", "M1", "M2", "D1", "D2"]
digits = [2, 0, 0, 4, 0, 8, 1, 1]
colors = ["royalblue", "royalblue", "royalblue", "royalblue",
          "mediumseagreen", "mediumseagreen",
          "darkorange", "darkorange"]

plt.figure(figsize=(9, 5))
bars = plt.bar(labels, digits, color=colors, edgecolor="black")

plt.title("Digits of My Birthday: August 11, 2004")
plt.xlabel("Position in the Date")
plt.ylabel("Digit Value")
plt.ylim(0, 9)

for bar, value in zip(bars, digits):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.15,
        str(value),
        ha="center",
        va="bottom"
    )

plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```
