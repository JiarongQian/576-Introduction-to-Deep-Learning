```python
import numpy as np

birthday = np.array([2, 0, 0, 4, 0, 8, 1, 1])
birthday_matrix = birthday.reshape(2, 4)

print("Birthday array:")
print(birthday)
print("Birthday matrix:")
print(birthday_matrix)
print("Array type:")
print(type(birthday))
print("Array shape:")
print(birthday.shape)
print("Matrix shape:")
print(birthday_matrix.shape)
print("Zero matrix:")
print(np.zeros((2, 4)))
print("One matrix:")
print(np.ones((2, 4)))
print("Constant matrix:")
print(np.full((2, 4), 8))
print("Identity matrix:")
print(np.eye(4))
```

```python
print("First element:")
print(birthday[0])

print("Last element:")
print(birthday[-1])

print("First four elements:")
print(birthday[:4])

print("First row:")
print(birthday_matrix[0, :])

print("Second row:")
print(birthday_matrix[1, :])

print("First two columns:")
print(birthday_matrix[:, :2])

print("Element in the second row and third column:")
print(birthday_matrix[1, 2])

print("Elements greater than one:")
print(birthday_matrix[birthday_matrix > 1])

print("Boolean index matrix:")
print(birthday_matrix > 1)
```

```python
integer_array = np.array([2, 0, 0, 4, 0, 8, 1, 1], dtype=np.int64)
float_array = np.array([2, 0, 0, 4, 0, 8, 1, 1], dtype=np.float64)
boolean_array = birthday > 0

print("Integer array:")
print(integer_array)

print("Integer data type:")
print(integer_array.dtype)

print("Floating-point array:")
print(float_array)

print("Floating-point data type:")
print(float_array.dtype)

print("Boolean array:")
print(boolean_array)

print("Boolean data type:")
print(boolean_array.dtype)
```

```python
x = birthday_matrix.astype(float)

y = np.array([
    [1., 2., 3., 4.],
    [5., 6., 7., 8.]
])

print("Element-wise addition:")
print(x + y)

print("Element-wise subtraction:")
print(x - y)

print("Element-wise multiplication:")
print(x * y)

print("Element-wise division:")
print(x / y)

print("Square root:")
print(np.sqrt(y))

print("Matrix multiplication:")
print(x @ y.T)

print("Sum of all elements:")
print(np.sum(x))

print("Sum of each column:")
print(np.sum(x, axis=0))

print("Sum of each row:")
print(np.sum(x, axis=1))

print("Transpose:")
print(x.T)
```

```python
row_offset = np.array([1., 2., 3., 4.])
row_result = x + row_offset

print("Original matrix:")
print(x)

print("Row offset:")
print(row_offset)

print("Result after row broadcasting:")
print(row_result)

column_offset = np.array([[10.], [20.]])
column_result = x + column_offset

print("Column offset:")
print(column_offset)

print("Result after column broadcasting:")
print(column_result)
```
